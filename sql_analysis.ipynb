Revenue by country

```python
query("""
    SELECT f.country,
           COUNT(DISTINCT f.freelancer_id) AS total_freelancers,
           ROUND(SUM(CAST(t.amount AS DECIMAL(10,2))), 2) AS total_revenue,
           ROUND(SUM(CAST(t.amount AS DECIMAL(10,2))) / COUNT(DISTINCT f.freelancer_id), 2) AS revenue_per_freelancer
    FROM clean_freelancer f
    JOIN clean_transactions t ON f.freelancer_id = t.freelancer_id
    WHERE LOWER(t.status) = 'success'
    GROUP BY f.country
    ORDER BY total_revenue DESC
""")
```

Cancellation rate

```python
query("""
    SELECT f.freelancer_id, f.name, f.country, f.primary_skill,
           COUNT(p.project_id) AS total_projects,
           SUM(CASE WHEN LOWER(p.status) = 'cancelled' THEN 1 ELSE 0 END) AS cancelled_projects,
           ROUND(100.0 * SUM(CASE WHEN LOWER(p.status) = 'cancelled' THEN 1 ELSE 0 END)
                 / COUNT(p.project_id), 1) AS cancel_rate_pct
    FROM clean_freelancer f
    JOIN clean_projects p ON f.freelancer_id = p.freelancer_id
    GROUP BY f.freelancer_id, f.name, f.country, f.primary_skill
    HAVING COUNT(p.project_id) >= 3
    ORDER BY 6 DESC
""")
```

Top rated freelancer per category

```python
query("""
    WITH ranked AS (
        SELECT f.name, f.country, p.category,
               ROUND(AVG(r.rating), 2) AS avg_rating,
               COUNT(r.review_id) AS total_reviews,
               RANK() OVER (
                   PARTITION BY p.category
                   ORDER BY AVG(r.rating) DESC
               ) AS rnk
        FROM clean_freelancer f
        JOIN clean_projects p ON f.freelancer_id = p.freelancer_id
        JOIN clean_reviews r ON p.project_id = r.project_id
        GROUP BY f.freelancer_id, f.name, f.country, p.category
    )
    SELECT category, name, country, avg_rating, total_reviews
    FROM ranked
    WHERE rnk = 1
    ORDER BY avg_rating DESC
""")
```

Freelancers with reviews below 2 stars

```python
query("""
    SELECT f.freelancer_id, f.name, f.country,
           COUNT(DISTINCT p.project_id) AS total_projects,
           SUM(CASE WHEN LOWER(p.status) = 'completed' THEN 1 ELSE 0 END) AS completed_projects,
           ROUND(AVG(r.rating), 2) AS avg_rating,
           MIN(r.rating) AS lowest_rating
    FROM clean_freelancer f
    JOIN clean_projects p ON f.freelancer_id = p.freelancer_id
    JOIN clean_reviews r ON p.project_id = r.project_id
    GROUP BY f.freelancer_id, f.name, f.country
    HAVING MIN(r.rating) < 2
    ORDER BY avg_rating ASC
""")
```

Running total of earnings per freelancer

```python
query("""
    SELECT f.name, t.payment_date, t.amount,
           ROUND(SUM(t.amount) OVER (
               PARTITION BY f.freelancer_id
               ORDER BY t.payment_date
               ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW
           ), 2) AS running_total
    FROM clean_freelancer f
    JOIN clean_transactions t ON f.freelancer_id = t.freelancer_id
    WHERE LOWER(t.status) = 'success'
    ORDER BY f.freelancer_id, t.payment_date
""")
```

Higher budget = better reviews?

```python
query("""
    SELECT
        CASE
            WHEN p.budget_usd < 500  THEN '1. Low (< $500)'
            WHEN p.budget_usd < 1500 THEN '2. Mid ($500-$1500)'
            WHEN p.budget_usd < 3000 THEN '3. High ($1500-$3000)'
            ELSE '4. Premium (> $3000)'
        END AS budget_tier,
        COUNT(p.project_id) AS total_projects,
        ROUND(AVG(r.rating), 2) AS avg_rating,
        ROUND(AVG(p.budget_usd), 2) AS avg_budget
    FROM clean_projects p
    JOIN clean_reviews r ON p.project_id = r.project_id
    GROUP BY budget_tier
    ORDER BY budget_tier
""")
```

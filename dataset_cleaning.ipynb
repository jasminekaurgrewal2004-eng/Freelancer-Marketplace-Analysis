```python
import pandas as pd
```

```python
import numpy as np
```

```python
df = pd.read_csv('freelancers.csv')
```

```python
df.head(10)
```

```python

df.info()
```

```python
df.describe()
```

```python
def mean():
    cols = ['hourly_rate_usd','years_experience','rating_avg','total_earnings_usd']
    for col in cols:
        df[col]= round(df[col].fillna(df[col].mean()),2)
    return df

mean()
```

```python
df_drop_all = df.dropna(how="all")
df_drop_all
```

```python
def exact_values():
    cols= ['profile_complete','verified']
    for col in cols:
        df[col] = df[col].replace({
        'True':1,
        "False":0,
        "true":1,
        "false":0,
        "YES":1,
        "NO":0,
        "yes":1,
        "no":0,
        "No":0,
        "Yes":1
    })
    return df
exact_values()
```

```python
df['verified'].info()
```

```python
df['verified'] = df['verified'].astype(int)
df['profile_complete'] = df['profile_complete'].astype(int)
```

```python
def string_conversion():
    cols = ['name','primary_skill','country','email']
    for col in cols:
        df[col]=df[col].astype(str)
    return df
string_conversion()
```

```python
df.info()
```

```python
def string_lowercase():
    cols = ['name','primary_skill','country','email']
    for col in cols:
        df[col]=df[col].str.strip().str.lower()
    return df
string_lowercase()
```

```python
df['join_date']= pd.to_datetime(df['join_date'])
```

```python
df.info()
```

```python
df=df.drop_duplicates()
```

```python
df.describe()
```

```python
# Q1 = 25 Q2=100
Q1 = df['hourly_rate_usd'].quantile(0.25)
Q3 = df['hourly_rate_usd'].quantile(0.75)
IQR = Q3-Q1

outliers = df[(df['hourly_rate_usd']<Q1-1.5*IQR)| (df['hourly_rate_usd']>Q3+1.5*IQR)]
outliers
```

```python
df['join_date']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.corr(numeric_only=True), annot=True)

plt.title('Correlation Heatmap')

plt.show()
```

```python
df.to_csv('clean_freelancer.csv',index = False)
```

2. PROJECTS DATASET

```python
import pandas as pd
```

```python
df = pd.read_csv('projects.csv')
```

```python
df.head(10)
```

```python
df.info()
```

```python
df['budget_usd'] = pd.to_numeric(df['budget_usd'],errors='coerce')
```

```python
def mean():
    cols = ['bids_received','budget_usd']
    for col in cols:
        df[col] = round(df[col].fillna(df[col].mean()),2)

    return df

mean()
```

```python
df.describe()
```

```python
def date_time():
    cols =['start_date','deadline']
    for col in cols:
        df[col]= pd.to_datetime(df[col],errors = 'coerce')
    return df

date_time()
```

```python
df.info()
```

```python
drop_null_rows = df.dropna(how='all')
drop_null_rows
```

```python
df['remote']= df['remote'].replace({
     'YES':1, 'Yes':1 , 'yes':1,
     'NO':0, 'No':0 , 'no':0,
     'TRUE':1, 'True':1, 'true':1,
     'FALSE':0 , 'False':0 , 'false':0
})
df['remote'] = df['remote'].astype(int)
df['remote']
```

```python
import matplotlib.pyplot as plt
```

```python
avg_budget = df.groupby('category')['budget_usd'].mean()

avg_budget.plot(kind='bar')
plt.title('Average Budget Per Job Category')
plt.ylabel('Average_Budget')
plt.show()
```

```python
duplicates = df.drop_duplicates()
```

```python
df.to_csv('clean_projects.csv',index=False)
```
